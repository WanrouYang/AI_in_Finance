```python
# Read file and retrieve lendingclub dataset
raw_data = pd.read_csv("/Users/yeung/Desktop/Path Academic (AI in Finance)/Final Project/LendingClub.csv", header = 0, skiprows = 1, low_memory=False)
raw_data
```

```python
# Read file and retrieve GDP dataset
gdp = pd.read_csv("/Users/yeung/Desktop/Path Academic (AI in Finance)/Final Project/GDP.csv")
gdp
```

```python
# Keep loans issued during Q2 2012
lending_club_data = raw_data[raw_data['issue_d'].isin(['Apr-2012', 'May-2012', 'Jun-2012'])].copy()
lending_club_data
```

```python
# Check and pick up what features/variables we are gonna use
lending_club_data.columns.tolist()
```

```python
# Add a new column called 'fico' by averaging the FICO scores from raw dataset
lending_club_data['fico'] = (
    lending_club_data['fico_range_low'] + lending_club_data['fico_range_high']
) / 2
```

```python
# Create binary default variable: Fully Paid = 0 (No default); Charged Off = 1 (Default)

lending_club_data = lending_club_data[
    lending_club_data['loan_status'].isin(['Fully Paid', 'Charged Off'])
].copy()

lending_club_data['default'] = lending_club_data['loan_status'].map({
    'Fully Paid': 0,
    'Charged Off': 1
})

lending_club_data[['loan_status','default']].head(10)
lending_club_data[['loan_status', 'default']].value_counts()
```

```python
# Define the borrower and loan characteristics that will be considered in the analysis and modeling stages.
# Final feature selection will be applied after the GDP data are merged.

candidate_features = lending_club_data[[
    # Credit information
    'fico',
    'fico_range_low',
    'fico_range_high',
    'dti',
    'delinq_2yrs',
    'inq_last_6mths',
    'pub_rec',
    'revol_util',
    'total_acc',

    # Borrower information
    'annual_inc',
    'emp_length',
    'home_ownership',

    # Loan information
    'loan_amnt',
    'int_rate',
    'term',
    'grade',
    'purpose',

    # Location information (for GDP merge)
    'addr_state',
    'issue_d'
]]

candidate_features
```

```python
# Add a new column called 'issue_year' for later dataset merging
lending_club_data['issue_d'].head()

lending_club_data['issue_year'] = pd.to_datetime(
    lending_club_data['issue_d'],
    format='%b-%Y'
).dt.year

lending_club_data[['issue_d', 'issue_year']].head()
```

```python
gdp.head()
```

```python
# Check dataset and pick up what features/variables we are gonna use
gdp.columns.tolist()
```

```python
# The dataset contains GDP information by different industries, so we only select "All industry total" to represent overall economic activity.
gdp = gdp[
    gdp['Description'] == 'All industry total'
].copy()

gdp.head()
```

```python
# Select the state name and the 2012 GDP value because the LendingClub sample is restricted to loans issued during Q2 2012.
# The GDP measure represents Total state GDP rather than GDP per capita.
# The values are reported in millions of current dollars.
gdp_2012 = gdp[
    ['GeoName', '2012']
].copy()
```

```python
# Rename the GDP column to indicate that it represents state GDP in 2012.
gdp_2012 = gdp_2012.rename(
    columns={'2012':'state_gdp_2012'}
)

gdp_2012.head()
```

```python
# Convert full state names in the GDP dataset into two-letter state abbreviations so that they match LendingClub's "addr_state" variable. 

# Example:
# California -> CA
# Massachusetts -> MA
# New York -> NY

def get_state_abbr(state_name):
    
    # Handle the District of Columbia explicitly.
    # BEA reports it as "District of Columbia", while LendingClub identifies borrowers from Washington, DC using "DC"
    if state_name == 'District of Columbia':
        return 'DC'
    
    # Use the "us" package to convert the remaining state names into their standard two-letter abbreviations
    state = us.states.lookup(state_name)
    
    if state:
        return state.abbr
    
    # Return None for observations that are not individual states, such as regional or national aggregates
    return None


# Apply the state-name conversion to the GDP dataset
gdp_2012['addr_state'] = gdp_2012['GeoName'].apply(
    get_state_abbr
)

# Check any geographic observations that could not be mapped
# These should only be non-state geographic aggregates
gdp_2012[
    gdp_2012['addr_state'].isna()
][['GeoName']]
```

```python
# Remove geographic observations that are not individual states or DC.
# These rows cannot be matched to LendingClub borrowers because borrowers
# are identified using state abbreviations.
gdp_2012 = gdp_2012.dropna(
    subset=['addr_state']
).copy()

# Keep only the variables needed for the LendingClub merge.
gdp_2012 = gdp_2012[
    ['addr_state', 'state_gdp_2012']
]

# Check the cleaned state-level GDP dataset.
gdp_2012.head()
```

```python
# Confirm the number of state-level GDP observations
# Ideally, this should represent the 50 states plus Washington, DC
print("Number of state/DC GDP observations:", len(gdp_2012))

# Check for duplicate state abbreviations
print(
    "Duplicate state abbreviations:",
    gdp_2012['addr_state'].duplicated().sum()
)
```

```python
# Merge 2012 state GDP into the LendingClub dataset using the borrower's state of residence ("addr_state") as the matching key
# A LEFT JOIN is used so that all LendingClub loans are retained, even if a GDP match is temporarily missing
merged_data = lending_club_data.merge(
    gdp_2012,
    on='addr_state',
    how='left'
)

# Preview the merged data.
merged_data.head()
```

```python
# Validate the GDP merge by comparing the number of loans before and after merging
# A successful left join should preserve the same number of LendingClub observations
print("Loans before GDP merge:", len(lending_club_data))
print("Loans after GDP merge:", len(merged_data))

# Count loans that did not receive a GDP value
print(
    "Loans with missing GDP:",
    merged_data['state_gdp_2012'].isna().sum()
)
```

```python
# Identify which borrower states, if any, failed to match with the state GDP dataset
# This allows us to fix mapping problems instead of simply deleting unmatched observations
unmatched_states = (
    merged_data.loc[
        merged_data['state_gdp_2012'].isna(),
        'addr_state'
    ]
    .value_counts()
)

unmatched_states
```

```python
# Select the final variables used for EDA, statistical analysis, and machine learning modeling. addr_state is retained for geographic analysis and validation, but will not be used as a predictor in the baseline-vs-GDP models.

final_data = merged_data[
[
    # Credit
    'fico',
    'dti',
    'delinq_2yrs',
    'inq_last_6mths',
    'pub_rec',
    'revol_util',
    'total_acc',

    # Borrower
    'annual_inc',
    'emp_length',
    'home_ownership',

    # Loan
    'loan_amnt',
    'int_rate',
    'term',
    'grade',
    'purpose',

    # Economic indicator
    'addr_state',
    'state_gdp_2012',

    # Target
    'default'
]
]
```

```python
# Final Dataset Validation

# Check the dimensions of the final modeling dataset.
print("Final dataset shape:", final_data.shape)

# Check missing values in each selected variable.
print("\nMissing values:")
print(
    final_data.isna()
    .sum()
    .sort_values(ascending=False)
)

# Check for duplicated observations.
print(
    "\nDuplicate rows:",
    final_data.duplicated().sum()
)

# Confirm the binary target distribution.
# 0 = Fully Paid
# 1 = Charged Off
print("\nDefault distribution:")
print(final_data['default'].value_counts())

print("\nDefault rate:")
print(final_data['default'].value_counts(normalize=True))

# Confirm that every borrower has matched GDP information.
print(
    "\nMissing GDP values:",
    final_data['state_gdp_2012'].isna().sum()
)

# Check how many borrower states are represented.
print(
    "\nNumber of borrower states:",
    final_data['addr_state'].nunique()
)
```
